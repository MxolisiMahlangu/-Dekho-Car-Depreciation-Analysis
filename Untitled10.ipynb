### 🚗 Questions for Car Depreciation Analysis

As a Data Analyst focusing on car value depreciation, you can explore the following key questions using this dataset:

1. **What is the average depreciation rate of cars?**
   - How does the ratio of `Selling_Price` to `Present_Price` change as the car's age increases?
   - What is the overall percentage of value retained after 1, 3, 5, and 10 years?

2. **How does fuel type affect depreciation?**
   - Do Petrol, Diesel, or CNG cars retain their value better over time?

3. **Does the transmission type impact the resale value?**
   - Do Manual cars depreciate faster or slower than Automatic cars?

4. **What is the relationship between mileage (`Kms_Driven`) and depreciation?**
   - Can we quantify the price drop per additional 10,000 kilometers driven?
   - Which has a stronger impact on depreciation: the age of the car (`Year`) or the mileage (`Kms_Driven`)?

5. **How does the seller type influence the selling price?**
   - Do Individual sellers experience steeper price drops compared to Dealers when looking at the same vehicle age?

6. **Do previous owners impact the depreciation curve?**
   - How much value does a car lose when it transitions from 1st owner to 2nd or 3rd owner (`Owner`)?

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calculate car age (assuming dataset baseline or using the max year in dataset as reference)
# Let's check the max year in the dataset first to use as a baseline
max_year = df['Year'].max()
df['Age'] = max_year - df['Year']

# 2. Calculate the Value Retention Ratio and Percentage
# Handling division by zero just in case Present_Price is 0
df['Value_Retained_Ratio'] = df['Selling_Price'] / df['Present_Price']
df['Value_Retained_Pct'] = df['Value_Retained_Ratio'] * 100

# 3. Plot Depreciation over Age
plt.figure(figsize=(10, 6))
sns.regplot(x='Age', y='Value_Retained_Pct', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Car Value Retention % vs. Age (Years)')
plt.xlabel('Age of Car (Years)')
plt.ylabel('Percentage of Original Value Retained (%)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(-0.5, df['Age'].max() + 0.5)
plt.ylim(0, 110)
plt.show()
```

```python
# Calculate the average retention percentage for specific years
years_of_interest = [1, 3, 5, 10]
retention_summary = {}

for year in years_of_interest:
    # Filter for cars of that specific age
    subset = df[df['Age'] == year]
    if not subset.empty:
        avg_retention = subset['Value_Retained_Pct'].mean()
        count = len(subset)
        retention_summary[year] = f"{avg_retention:.2f}% (based on {count} cars)"
    else:
        retention_summary[year] = "No data available for this age"

# Also show overall averages using a rolling or grouped approach for a general trend
grouped_retention = df.groupby('Age')['Value_Retained_Pct'].agg(['mean', 'count']).reset_index()

print("--- Retained Value After Specific Years ---")
for yr, val in retention_summary.items():
    print(f"After {yr} Year(s): {val}")

print("\n--- Full Retention Table by Age ---")
display(grouped_retention.sort_values(by='Age'))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by Fuel_Type and Age to see retention trends
fuel_retention = df.groupby(['Fuel_Type', 'Age'])['Value_Retained_Pct'].mean().reset_index()

# Create the line plot
plt.figure(figsize=(10, 6))
sns.lineplot(x='Age', y='Value_Retained_Pct', hue='Fuel_Type', marker='o', data=fuel_retention, linewidth=2)
plt.title('Value Retention % over Time by Fuel Type')
plt.xlabel('Age of Car (Years)')
plt.ylabel('Average Value Retained (%)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(-0.5, df['Age'].max() + 0.5)
plt.ylim(0, 110)
plt.legend(title='Fuel Type')
plt.show()
```

```python
# Calculate overall summary statistics for fuel types
overall_fuel_summary = df.groupby('Fuel_Type').agg(
    Avg_Value_Retained_Pct=('Value_Retained_Pct', 'mean'),
    Avg_Present_Price=('Present_Price', 'mean'),
    Avg_Selling_Price=('Selling_Price', 'mean'),
    Car_Count=('Car_Name', 'count')
).reset_index()

print("--- Overall Performance by Fuel Type ---")
display(overall_fuel_summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# 1. Plot Mileage vs Value Retention
plt.figure(figsize=(10, 6))
sns.regplot(x='Kms_Driven', y='Value_Retained_Pct', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'purple'})
plt.title('Car Value Retention % vs. Mileage (Kms Driven)')
plt.xlabel('Kilometers Driven')
plt.ylabel('Percentage of Original Value Retained (%)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# 2. Quantify drop per 10,000 km using linear regression
X = sm.add_constant(df['Kms_Driven'])
y = df['Value_Retained_Pct']
model_kms = sm.OLS(y, X).fit()

# Slope represents change in percentage per 1 km
slope_per_km = model_kms.params['Kms_Driven']
slope_per_10k_km = slope_per_km * 10000

print(f"--- Mileage Impact Quantification ---")
print(f"On average, for every additional 10,000 kilometers a car is driven,")
print(f"it loses {abs(slope_per_10k_km):.2f}% of its original value.")
```

```python
# 3. Compare correlation strength of Age vs Kms_Driven with respect to Value_Retained_Pct
correlation_matrix = df[['Value_Retained_Pct', 'Age', 'Kms_Driven']].corr()

print("--- Correlation Matrix ---")
display(correlation_matrix)

print("\n--- Key Takeaway ---")
age_corr = correlation_matrix.loc['Value_Retained_Pct', 'Age']
kms_corr = correlation_matrix.loc['Value_Retained_Pct', 'Kms_Driven']

print(f"Correlation between Value Retention and Age: {age_corr:.4f}")
print(f"Correlation between Value Retention and Mileage: {kms_corr:.4f}")

if abs(age_corr) > abs(kms_corr):
    print("Conclusion: Car AGE has a stronger negative impact on depreciation than mileage.")
else:
    print("Conclusion: MILEAGE (Kms Driven) has a stronger negative impact on depreciation than age.")
```
